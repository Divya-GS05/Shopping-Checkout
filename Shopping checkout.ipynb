```python
Price = 600
Quantity = 3
has_coupon = True
Budget = 2000

#Subtotal
Sub_Total = Price*Quantity
print("Sub total without discount: ", Sub_Total)

#Discount
Discount_Eligibility = Sub_Total>=1000 and has_coupon
if Discount_Eligibility:
  Total_Amount = Sub_Total-100
  print("Discount is applied")


#Customer Budget
Customer_Budget = Total_Amount<=Budget

#Results
print("Whether customer meets the budget: ",Customer_Budget)
print("Total Amount: ", Total_Amount)


```
